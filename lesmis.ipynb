```python
import networkx as nx
import matplotlib.pyplot as plt
```

```python
path = "/Users/taira_no_mac/Documents/GitHub/DeepWalkA/data/lesmis.gml"
gml = nx.read_gml(path)
```

```python
plt.figure(figsize=(36, 30))
pos = nx.spring_layout(gml)
nx.draw_networkx(gml,pos, font_size = 8, node_size = 100,with_labels = False,width = 0.5)
edge_labels = nx.get_edge_attributes(gml, 'value')
nx.draw_networkx_edge_labels(gml, pos, edge_labels = edge_labels,font_size = 10)
plt.show()
```
